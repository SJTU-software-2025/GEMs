```python
"""
BRENDA: TURNOVER NUMBER [1/s] 	SUBSTRATE
Sabio-RK:

https://cobrapy.readthedocs.io/en/latest/media.html:
fluxes have the unit mmol / [gDW h] (concentration per gram dry weight of cells and hour)

1kDa表示相对分子质量为1000的分子
酶的分子量为 50kDa = 50000g/mol(典型值)
假设总酶量占细胞干重的 40% (即 0.4g酶/gDW)。
总酶池约束为 0.008 mmol酶/gDW
"""

for reaction in model.reactions:
    # 每个反应对应一个酶
    # ub设置为总酶池约束
    enzyme_var = model.problem.Variable(f"Enzyme_{reaction.id}", lb=0, ub=8)
    model.add_cons_vars(enzyme_var)

    # 中位数约0.5（单位：1/s）
    kcat = 1800 # (单位：1/h)
    constraint_max = model.problem.Constraint(
        reaction.flux_expression - kcat * enzyme_var,
        ub=0,  # v_i - kcat*E_i ≤ 0 → v_i ≤ kcat*E_i
        name=f"enzyme_constraint_{reaction.id}_max"
    )
    constraint_min = model.problem.Constraint(
        reaction.flux_expression + kcat * enzyme_var,
        lb=0,  # v_i + kcat*E_i ≥ 0 → v_i ≥ -kcat*E_i
        name=f"enzyme_constraint_{reaction.id}_min"
    )
    model.add_cons_vars([constraint_max, constraint_min])

total_protein_content = 8  # 单位：0.008mmol/gDW（假设值）× 1000 = 8mmol/gDW (弥补求解器精度不足)
enzyme_vars = [ var for var in model.variables if var.name.startswith("Enzyme_")]

total_enzyme_expr = sum(enzyme_vars)
total_enzyme_constraint = model.problem.Constraint(
    total_enzyme_expr,
    lb=0,
    ub=total_protein_content,
    name="total_enzyme_constraint"
)
model.add_cons_vars(total_enzyme_constraint)

solution = model.optimize()
print(solution.fluxes)
print(model.summary())
```

```python
def find_essential_single_reaction(model, production_threshold=0.4):
    """
    Find essential single reaction
    param production_threshold: The minimum production change threshold to consider a reaction essential.
    return: A list of essential reactions.
    """
    essential_reactions = []
    objective_value = model.optimize().objective_value
    single_reaction_deletion_results = pd.read_csv('../data/single_reaction_deletion_results.csv')
    for _, row in single_reaction_deletion_results.iterrows():
        if row['status'] != 'optimal':
            continue
        if abs((row['growth'] - objective_value) / objective_value) >= production_threshold:
            essential_reactions.append(row['ids'])
    return essential_reactions

essential_reactions = find_essential_single_reaction(model)
print(f"Essential reactions: {essential_reactions}")
```

```python
from bioservices import KEGG
import pandas as pd
import re
from tqdm import tqdm

starttime = datetime.now()

# 初始化KEGG服务
k = KEGG()

# 读取数据
chr_kegg_metabolism = pd.read_csv('../data/NRRL_1/7_Annotation/chr.kegg.metabolism.csv')

# 预先创建新列
chr_kegg_metabolism["R number"] = None

for index, row in tqdm(chr_kegg_metabolism.iterrows(), total=len(chr_kegg_metabolism),  desc="Processing KEGG Entries"):
    ko_number = row['KO number']
    try:
        # 调用API
        data = k.get(ko_number)
        
        if not data: # 处理空响应
            raise ValueError(f"Empty response for {ko_number}")
            
        dict_data = k.parse(data)
        
        # 提取反应编号（带异常处理）
        reaction_numbers = []
        if "REACTION" in dict_data:
            reactions = dict_data["REACTION"]
            if isinstance(reactions, str):
                reactions = [reactions]
            # 严格匹配 RXXXXX 格式
            reaction_numbers = [
                item.strip() for item in reactions
                if re.fullmatch(r'R\d{5}', item.strip()) # 去除空格后匹配
            ]
        
        # 将列表保存到新列（用分号分隔）
        chr_kegg_metabolism.at[index, "R number"] = "; ".join(reaction_numbers)
        
        # 写入日志
        write_log(f"Processed {ko_number}: Found {len(reaction_numbers)} reactions")
    
    except Exception as e:
        error_msg = f"Error processing {ko_number}: {str(e)}"
        write_log(error_msg)
        chr_kegg_metabolism.at[index, "R number"] = "Error"
        continue

chr_kegg_metabolism.to_csv(
    '../data/NRRL_1/7_Annotation/chr.kegg_metabolism_with_reactions.csv', 
    index=False
)
write_log("Processing completed. Results saved.")

endtime = datetime.now()
print({endtime - starttime})
```

```python
# chr.kegg.reactions.csv
from bioservices import KEGG
import pandas as pd
from datetime import datetime
import os
from tqdm import tqdm
import time

starttime = datetime.now()

# 定义日志文件路径
log_dir = '../data/NRRL_1/7_Annotation/logs'
log_file = os.path.join(log_dir, 'kegg_api_processing2.log')
def write_log(message, log_path=log_file):
    """将带时间戳的日志写入文件和打印到控制台"""
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    log_entry = f"[{timestamp}] {message}"
    
    # 写入日志文件
    with open(log_path, 'a', encoding='utf-8') as f:
        f.write(log_entry + '\n')
    
    # 同时打印到控制台
    # print(log_entry)
    
# pd.DataFrame: Reaction ID	Reaction (Long)	Reaction (Short)	Reversible	Enzyme
# 创建存储结果的DataFrame
reaction_df = pd.DataFrame(columns=[
    "Reaction ID", 
    "Reaction (Long)", 
    "Reaction (Short)", 
    "Reversible",
    "Enzyme"
])

# 初始化KEGG服务
k = KEGG()

# 读取带反应编号的数据
chr_kegg_metabolism_with_reactions = pd.read_csv(
    '../data/NRRL_1/7_Annotation/chr.kegg_metabolism_with_reactions.csv',
    dtype={'R number': str}
).fillna({'R number': ''})

processed_reactions = set()  # 记录已处理过的反应，避免重复获取

for _, row in tqdm(chr_kegg_metabolism_with_reactions.iterrows(), total=len(chr_kegg_metabolism_with_reactions), desc="Processing KEGG reactions"):
    r_number = row['R number'].strip()
    if not r_number:  # 跳过空字符串
        continue
    r_list = [r_id.strip() for r_id in r_number.split(';')]
    for r_id in r_list:
         # 跳过已处理或空值
        if not r_id or r_id in processed_reactions:
            continue
        try: 
            # API速率控制 (每秒最多3次)
            # time.sleep(0.34)
            # 获取反应数据
            data = k.get(r_id)
            if not data:
                raise ValueError(f"Empty response for {r_id}")
            dict_data = k.parse(data)
            # 解析数据 
            reaction_long = dict_data.get('DEFINITION', '')
            reaction_short = dict_data.get('EQUATION', '')
            # 根据方程中的箭头类型判断可逆性
            reversible = '<=>' in reaction_short  
            # 提取酶编号
            enzymes = dict_data.get('ENZYME', [])            
            # 添加到结果DataFrame
            new_row = pd.DataFrame([{
                "Reaction ID": r_id,
                "Reaction (Long)": reaction_long,
                "Reaction (Short)": reaction_short,
                "Reversible": reversible,
                "Enzyme": ";".join(enzymes)
            }])
            reaction_df = pd.concat([reaction_df, new_row], ignore_index=True)
            processed_reactions.add(r_id)
            # 写入日志
            write_log(f"Processed {r_id} successfully")
        except Exception as e:
            error_msg = f"Error processing {r_id}: {str(e)}"
            write_log(error_msg)
            # 记录错误条目
            error_row = pd.DataFrame([{
                "Reaction ID": r_id,
                "Reaction (Long)": "ERROR",
                "Reaction (Short)": "ERROR",
                "Reversible": "ERROR",
                "Enzyme": "ERROR"
            }])
            reaction_df = pd.concat([reaction_df, error_row], ignore_index=True)
            continue
# 保存最终结果
reaction_df.to_csv('../data/NRRL_1/7_Annotation/chr.kegg.reactions.csv', index=False)

endtime = datetime.now()
print(endtime - starttime)
```

```python
from bioservices import KEGG
import pandas as pd
import os
from datetime import datetime
from tqdm import tqdm

starttime = datetime.now()

# 定义日志文件路径
log_dir = '../data/NRRL_1/7_Annotation/logs'
log_file = os.path.join(log_dir, 'kegg_api_processing3.log')
def write_log(message, log_path=log_file):
    """将带时间戳的日志写入文件和打印到控制台"""
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    log_entry = f"[{timestamp}] {message}"
    
    # 写入日志文件
    with open(log_path, 'a', encoding='utf-8') as f:
        f.write(log_entry + '\n')
    
    # 同时打印到控制台
    # print(log_entry)
    
# pd.DataFrame: Metabolite ID	Name Formula/Composition
# 创建存储结果的DataFrame
metabolite_df = pd.DataFrame(columns=[
    "Metabolite ID", 
    "Name",
    "Formula/Composition",
    "Compartment"
])

processed_metabolites = set()  # 记录已处理过的代谢物，避免重复获取

reactions_df = pd.read_csv('../data/NRRL_1/7_Annotation/chr.kegg.reactions.csv')

k = KEGG()

for _, row in tqdm(reactions_df.iterrows(), total=len(reactions_df), desc="Processing KEGG Metabolites"):
    reaction_short = row['Reaction (Short)']
    reactants_products = []
    # 查询各个反应物的分子式
    reactants = reaction_short.split(' <=> ')[0].split(' + ')
    products = reaction_short.split(' <=> ')[1].split(' + ')
    for reactant in reactants:
        reactant = reactant.split(' ')[-1]  # 提取代谢物ID
        reactants_products.append(reactant)
    for product in products:
        product = product.split(' ')[-1]
        reactants_products.append(product)
    for metabolite_id in reactants_products:
        metabolite_id = metabolite_id.strip()
        if not metabolite_id or metabolite_id in processed_metabolites:
            continue
        try:
            # 获取代谢物数据
            data = k.get(metabolite_id)
            if not data:
                raise ValueError(f"Empty response for {metabolite_id}")
            dict_data = k.parse(data)
            name = formula = ''
            # 解析数据 
            if metabolite_id.startswith('C'):
                name = dict_data.get('NAME', '')[0]
                formula = dict_data.get('FORMULA', '')
            if metabolite_id.startswith('G'):
                name = dict_data.get('ENTRY', '').split()[1]
                formula = dict_data.get('COMPOSITION', '')
            # 添加到结果DataFrame
            new_row = pd.DataFrame([{
                "Metabolite ID": metabolite_id,
                "Name": name,
                "Formula/Composition": formula
                "Compartment": 'Cytosol'
            }])
            metabolite_df = pd.concat([metabolite_df, new_row], ignore_index=True)
            processed_metabolites.add(metabolite_id)
            # 写入日志
            write_log(f"Processed {metabolite_id} successfully")
        except Exception as e:
            error_msg = f"Error processing {metabolite_id}: {str(e)}"
            write_log(error_msg)
            # 记录错误条目
            error_row = pd.DataFrame([{
                "Metabolite ID": metabolite_id,
                "Name": "ERROR",
                "Formula/Composition": "ERROR"
            }])
            metabolite_df = pd.concat([metabolite_df, error_row], ignore_index=True)
            continue

metabolite_df.to_csv('../data/NRRL_1/7_Annotation/chr.kegg.metabolites.csv', index=False)
endtime = datetime.now()
print(endtime - starttime)
```
